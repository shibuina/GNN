```python
import pandas as pd
import networkx as nx
from networkx.algorithms.community import k_clique_communities
from itertools import combinations

data = pd.read_csv("movielens_1m.csv")

like_threshold = 4
liked = data[data['rating'] >= like_threshold]

G = nx.Graph()
movies = liked['movieId'].unique()
G.add_nodes_from(movies)

for user, group in liked.groupby("userId"):
    movie_list = group['movieId'].tolist()
    if len(movie_list) >= 2:
        for movie1, movie2 in combinations(movie_list, 2):
            if G.has_edge(movie1, movie2):
                G[movie1][movie2]['weight'] += 1
            else:
                G.add_edge(movie1, movie2, weight=1)

print("Graph constructed:")
print(f"Nodes: {G.number_of_nodes()}, Edges: {G.number_of_edges()}")

k = 3
communities = list(k_clique_communities(G, k))
print(f"Found {len(communities)} communities using {k}-clique percolation.")

for idx, community in enumerate(communities[:10]):
    print(f"Community {idx+1} (size {len(community)}): {sorted(community)}")
```
